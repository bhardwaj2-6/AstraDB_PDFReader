```python
from typing_extensions import Concatenate
raw_text = ''
for i, page in enumerate(pdfreader.pages):
  content = page.extract_text()
  if content:
    raw_text += content
```

```python
cassio.init(token=ASTRA_DB_APPLICATION_TOKEN,database_id=ASTRA_DB_ID)
```

```python
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
model_name = "gemma2-9b-it"
llm = ChatGroq(
    temperature=0.1,
    model_name=model_name,  # Using model_name parameter
    api_key=GROQ_API_KEY
)
```

```python
astra_vector_store = Cassandra(
    embedding=embedding_model,
    table_name="qa_mini_demo",
    session=None,
    keyspace=None
)
```

```python
from langchain.text_splitter import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator = "/n",
    chunk_size= 500,
    chunk_overlap= 50,
    length_function=len
)
texts = text_splitter.split_text(raw_text)
```

```python
texts[:50]
```

```python
# Call add_texts before wrapping with VectorStoreIndexWrapper
astra_vector_store.add_texts(texts[:50])
print("inserted %i headlines." % len(texts[:50]))
# Now wrap with VectorStoreIndexWrapper
astra_vector_store = VectorStoreIndexWrapper(vectorstore=astra_vector_store)
```

```python
first_question = True
while True:
  if first_question:
    query_text= input("\nEnter your question(or type 'quite' to exit: )").strip()
  else:
    query_text = input("\nwhat's your next question (or type 'quite' to exit:)").strip()
  if query_text.lower() == 'quite':
    break

  if query_text == "":
    continue

  first_question = False
  print("\nQUESTION: \"%s\"" % query_text)
  # The following lines were incorrectly indented
  answer = astra_vector_store.query(query_text, llm=llm).strip() # Corrected indentation
  print("ANSWER: \"%s\"\n" % answer) # Corrected indentation

  print("FIRST DOCUMENTS BY RELEVANCE:") # Corrected indentation
  # Use the underlying vectorstore for similarity_search_with_score
  for doc, score in astra_vector_store.vectorstore.similarity_search_with_score(query_text, k=4): # Corrected indentation
      print("    [%0.4f] \"%s ...\"" % (score, doc.page_content[:84])) # Corrected indentation
```
